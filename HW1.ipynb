```python
import polars as pl
```

```python
# 1. Считывание датасета из файла train.csv
df = pl.read_csv('train.csv')
print("1. Датасет успешно загружен")
print(f"Размер датасета: {df.shape}")
print()
```

```python
# 2. Основная информация о датасете
print("2. Основная информация о датасете:")
print("Типы данных:")
print(df.schema)
print()
```

```python
print("Число пропусков по столбцам:")
null_counts = df.null_count()
print(null_counts)
print()
```

```python
print("Основные статистические характеристики:")
print(df.describe())
print()
```

```python
# 3. Количество пассажиров каждого класса
print("3. Количество пассажиров каждого класса:")
pclass_count = df.group_by('Pclass').agg(
    pl.count().alias('count')
).sort('Pclass')
print(pclass_count)
print()
```

```python
# 4. Количество выживших мужчин и женщин
print("4. Количество выживших мужчин и женщин:")
survived_by_sex = df.filter(pl.col('Survived') == 1).group_by('Sex').agg(
    pl.count().alias('survived_count')
)
print(survived_by_sex)
print()

# Альтернативный вариант - таблица сопряженности
print("Детальная статистика по полу и выживанию:")
survival_table = df.group_by(['Sex', 'Survived']).agg(
    pl.count().alias('count')
)
print(survival_table)
print()
```

```python
# 5. Пассажиры старше 44 лет
print("5. Пассажиры старше 44 лет:")
older_passengers = df.filter(pl.col('Age') > 44)
print(f"Количество пассажиров старше 44 лет: {older_passengers.height}")
print(older_passengers)
```

## Задание 2

```python
import pandas as pd
import bottleneck as bn
import numpy as np
```

```python
df = pd.read_csv('train.csv')
print("Датасет успешно загружен")
print(f"Размер датасета: {df.shape}")
print()
```

```python
# 1. Средний возраст пассажиров и стандартное отклонение с помощью bottleneck
# Используем bottleneck для вычисления среднего и стандартного отклонения
mean_age = bn.nanmean(df['Age'].values)  # nanmean игнорирует NaN значения
std_age = bn.nanstd(df['Age'].values)    # nanstd игнорирует NaN значения
```

```python
print(f"Средний возраст пассажиров: {mean_age:.2f}")
print(f"Стандартное отклонение возраста: {std_age:.2f}")
```

```python
# Для сравнения - стандартные методы pandas
pandas_mean = df['Age'].mean()
pandas_std = df['Age'].std()
print(f"\nДля сравнения (pandas):")
print(f"Средний возраст: {pandas_mean:.2f}")
print(f"Стандартное отклонение: {pandas_std:.2f}")
print()
```

```python
# 2. Создание нового столбца Fare_new с векторизованными операциями
# Векторизованный подход с использованием np.where
df['Fare_new'] = np.where(df['Pclass'].isin([1, 2]),
                         df['Fare'] * 1.3,
                         df['Fare'] * 1.1)
```

```python
print("Первые 10 строк с новым столбцом Fare_new:")
print(df[['Pclass', 'Fare', 'Fare_new']].head(10))
print()
```

```python
# Проверка результатов по классам
print("Средняя стоимость билета по классам:")
fare_stats = df.groupby('Pclass').agg({
    'Fare': 'mean',
    'Fare_new': 'mean'
}).round(2)
print(fare_stats)
```

## Задание 3

```python
# Считываем датасет
df = pd.read_csv('Housing.csv')
```

```python
print("Исходный Датафрейм ")
print(f"Размер датафрейма: {df.shape}")
print(f"Потребление памяти до оптимизации: {df.memory_usage(deep=True).sum() / 1024 ** 2:.2f} MB")
```

```python
print("\nИнформация о данных")
df.info()
```

```python
print("\nТипы данных до оптимизации")
print(df.dtypes)
```

```python
print("\nУникальные значения категориальных переменных")
categorical_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating',
                      'airconditioning', 'prefarea', 'furnishingstatus']
for col in categorical_columns:
    print(f"{col}: {df[col].unique()}")
```

```python
print("\nДиапазоны числовых переменных")
numeric_columns = ['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']
for col in numeric_columns:
    print(f"{col}: min={df[col].min()}, max={df[col].max()}, dtype={df[col].dtype}")
```

```python
# Оптимизация типов данных
# Сохраняем копию для сравнения
df_optimized = df.copy()
```

```python
# Оптимизация числовых столбцов
df_optimized['price'] = df_optimized['price'].astype('int32')
df_optimized['area'] = df_optimized['area'].astype('int16')
df_optimized['bedrooms'] = df_optimized['bedrooms'].astype('int8')
df_optimized['bathrooms'] = df_optimized['bathrooms'].astype('int8')
df_optimized['stories'] = df_optimized['stories'].astype('int8')
df_optimized['parking'] = df_optimized['parking'].astype('int8')
```

```python
# Оптимизация категориальных столбцов
categorical_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating',
                      'airconditioning', 'prefarea', 'furnishingstatus']
for col in categorical_columns:
    df_optimized[col] = df_optimized[col].astype('category')
```

```python
print("\nСравнение памяти до и после оптимизации")
memory_before = df.memory_usage(deep=True).sum() / 1024 ** 2
memory_after = df_optimized.memory_usage(deep=True).sum() / 1024 ** 2
reduction = (memory_before - memory_after) / memory_before * 100
print(f"Память до оптимизации: {memory_before:.2f} MB")
print(f"Память после оптимизации: {memory_after:.2f} MB")
print(f"Экономия памяти: {reduction:.1f}%")

print("\nТипы данных после оптимизации")
print(df_optimized.dtypes)
```

```python
print("\nДетальное сравнение по столбцам")
for col in df.columns:
    mem_before = df[col].memory_usage(deep=True) / 1024
    mem_after = df_optimized[col].memory_usage(deep=True) / 1024
    reduction_col = (mem_before - mem_after) / mem_before * 100 if mem_before > 0 else 0

    print(f"{col:20} | {str(df[col].dtype):10} -> {str(df_optimized[col].dtype):15} | "
          f"{mem_before:6.1f} KB -> {mem_after:5.1f} KB | Экономия: {reduction_col:5.1f}%")
```

```python
# Проверяем, что данные не изменились
print("\nПроверка сохранения данных")
print("Первые 5 строк после оптимизации:")
print(df_optimized.head())
```

```python
print("\nУникальные значения категориальных переменных после оптимизации:")
for col in categorical_columns:
    print(f"{col}: {df_optimized[col].unique()}")
```
